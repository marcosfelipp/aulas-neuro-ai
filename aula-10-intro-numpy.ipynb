## Por que NumPy?

Até aqui, dobrar um milhão de números era um laço. Agora é uma conta só. Rode as duas
células e compare — o `%timeit` repete a medida várias vezes e mostra a média.

```python
lista = list(range(1_000_000))

%timeit [x * 2 for x in lista]
```

```python
a = np.arange(1_000_000)

%timeit a * 2
```

A diferença costuma ficar entre 20 e 100 vezes. E não é só tempo — é memória também:



A lista guarda **ponteiros** para um milhão de objetos espalhados pela memória: cada
`x * 2` precisa ir buscar um objeto, descobrir o tipo dele e criar outro. O array é **um
bloco contíguo** de números do mesmo tipo, e a conta roda sobre o bloco inteiro em código
compilado. As duas vantagens saem da mesma decisão de projeto.

### Operações elemento a elemento

É isto que substitui o laço: a conta vale para o array inteiro de uma vez, e o resultado é
outro array do mesmo tamanho.

```python
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])
```

Repare na última linha: comparar também é elemento a elemento, e o resultado é um array de
`True`/`False`. É disso que a próxima seção vive.

### Máscara booleana

A célula do `a[a<8]` lá em cima já usou isto. Vale ver devagar, porque é o `if` dentro do
laço, aposentado.



```python
a = np.array([3, 9, 1, 12, 7, 15])

print(a > 5)            # a máscara: True/False, do tamanho do array
print(a[a > 5])         # só os que passaram
print((a > 5).sum())    # quantos passaram — True vale 1 numa soma
print(a[a > 5].mean())  # a média só dos que passaram
```

Em Python puro, isso era um laço com `if` e `.append()`. E a contagem sai de graça porque
`True` vale `1` quando entra numa soma.
