## O que ficou

| Conceito | Onde apareceu |
|---|---|
| `def` | `def duracao(tempo):` — dar nome a um pedaço de código |
| Parâmetro | o que muda a cada chamada, sem copiar a linha |
| Dois parâmetros | `passou_do_limiar(valor, limiar)` — separados por vírgula |
| `return` | **entrega** o valor e encerra a função |
| Devolver `True`/`False` | função não devolve só número |
| Uma função chamando outra | `taxa_de_disparo` usa `duracao`; `relatorio` usa as três |

E a ideia que vale mais que a sintaxe: **toda vez que você copiar e colar código mudando
um detalhe, esse detalhe quer virar parâmetro.**
