````python
import numpy as np
````

## Fabricando a sequência
`np.arange` : https://numpy.org/doc/stable/reference/generated/numpy.arange.html <br>
`np.linspace` : https://numpy.org/doc/stable/reference/generated/numpy.linspace.html

As duas geram uma sequência pronta. A diferença é **o que você escolhe**: no `arange`, o
passo; no `linspace`, a quantidade de pontos.

````python
print(np.arange(0, 12, 2))      # do 0 ao 12, de 2 em 2
print(np.linspace(0, 12, 5))    # 5 pontos igualmente espaçados entre 0 e 12
````



Repare no fim da linha: o `arange` **deixa o 12 de fora** — é o mesmo comportamento do
fatiamento, onde `lista[0:3]` não traz o item 3. Já o `linspace` **inclui as duas pontas**,
porque o que ele promete é a quantidade de pontos.

````python
print(np.arange(13))              # 0 a 12 — números inteiros
print(np.linspace(0, 12, 13))     # os mesmos 0 a 12 — mas com ponto decimal

print(np.arange(13).dtype)        # int64
print(np.linspace(0, 12, 13).dtype)   # float64
````

Mesmos treze números, tipos diferentes. O `linspace` sempre devolve `float`, porque ele
divide o intervalo — e dividir raramente dá número redondo.

## Escrevendo dentro do array

Até aqui o colchete só serviu para **ler**. Ele também escreve — basta pôr do lado esquerdo
do `=`. E tudo que servia para escolher o que ler serve para escolher o que escrever.



````python
v = np.zeros(5)

v[0] = 18          # uma posição
print(v)

v[1:3] = 10        # uma fatia inteira, de uma vez só
print(v)

w = np.array([18.0, 10, 10, 3, 1])
w[w < 5] = 0       # por máscara: todos os que passam no teste
print(w)
````

A primeira linha é a mesma escrita que você já usava em dicionário — `contagem[s] = ...`,
`exames[paciente] = ...`. Muda o recipiente, não a gramática.

A última é a máscara da aula 10 do outro lado do `=`: antes ela escolhia o que **ler**,
agora escolhe o que **escrever**. Em Python puro essa linha seria um laço com `if` dentro.
