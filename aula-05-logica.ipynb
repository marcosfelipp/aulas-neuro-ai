### As cinco primeiras, junto com o professor

````python
print("True or True and False     ->", True or True and False)
print("not 5 > 3                  ->", not 5 > 3)
print("(True or True) and False   ->", (True or True) and False)
print("5 <= 7 <= 10               ->", 5 <= 7 <= 10)
print("not (True and False)       ->", not (True and False))
````

### Agora as cinco suas

**Escreva a sua previsão no comentário de cada linha antes de rodar.** Depois rode e veja
quantas você acertou.

````python
# previsão: ____
print("False or not False            ->", False or not False)

# previsão: ____
print("3 > 2 > 1                     ->", 3 > 2 > 1)

# previsão: ____
print("not True and False            ->", not True and False)

# previsão: ____
print("True and not (False or True)  ->", True and not (False or True))

# previsão: ____
print("10 > 5 or 1/0 > 0             ->", 10 > 5 or 1/0 > 0)
````

<details>
<summary><b>A última não deveria ter quebrado. Por que não quebrou?</b></summary>

`1/0` é uma divisão por zero — sozinha, ela estoura `ZeroDivisionError`.

Mas o `or` é **preguiçoso**: ele avalia da esquerda para a direita e **para no primeiro
verdadeiro**. Como `10 > 5` já é `True`, o resto nunca chega a ser executado.

Isso tem nome — *curto-circuito* — e serve de proteção:

```python
if len(valores) > 0 and soma / len(valores) > 10:
```

O `and` também para no primeiro falso. Se a lista estiver vazia, a divisão nunca acontece.

</details>

<details>
<summary><b>Abra só depois de rodar o teste acima</b></summary>

**Se a última linha quebrou, você escreveu sem parênteses.**

```python
18 <= idade <= 65 and escore >= 10 or usa_medicacao      # errado
18 <= idade <= 65 and (escore >= 10 or usa_medicacao)    # certo
```

O `and` tem precedência sobre o `or`, então o Python leu a primeira versão assim:

```python
(18 <= idade <= 65 and escore >= 10) or usa_medicacao
```

Com `idade=70`, `escore=5`, `usa_medicacao=True`, a primeira parte dá `False`, mas o
`or usa_medicacao` salva e devolve `True` — **um paciente de 70 anos entrou num estudo
que vai de 18 a 65.**

Nos outros seis casos as duas versões concordam. Foi preciso um teste específico para
achar. É assim que esse bug vive em produção.

</details>
