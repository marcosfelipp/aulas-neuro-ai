## Parte 3 · Dicionários

Seis funções, da mais curta para a mais longa. As três primeiras saem em uma ou duas
linhas; as três últimas são o mesmo laço, mudando só o que entra no acumulador.

### 7 · O escore do paciente

Devolva o escore do paciente. Quem não está no dicionário vale `0`.

````python
def escore(exames, paciente):
    """Devolve o escore de `paciente` em `exames`. Quem não está lá vale `0`."""
````

````python
exames = {"P001": 17, "P002": 4}

assert escore(exames, "P001") == 17
assert escore(exames, "P002") == 4
assert escore(exames, "P999") == 0

print("ok")
````

<details>
<summary><b>Uma linha — e qual delas</b></summary>

`exames[paciente]` quebra no terceiro teste: `KeyError: 'P999'`. O `.get` devolve a
reserva que você mandar:

```python
return exames.get(paciente, 0)
```

É o `[]` cobra, `.get()` perdoa do slide. Sem o `, 0`, o retorno seria `None` — e `None`
não serve para somar depois.

</details>

### 8 · Registrar um escore

Guarde `valor` no paciente e devolva o dicionário. Se o paciente já estiver lá, o valor
novo substitui o antigo.

````python
def registra(exames, paciente, valor):
    """Guarda `valor` em `paciente` e devolve `exames`."""
````

````python
assert registra({}, "P001", 17)           == {"P001": 17}
assert registra({"P001": 17}, "P002", 4)  == {"P001": 17, "P002": 4}
assert registra({"P001": 17}, "P001", 21) == {"P001": 21}     # substitui

print("ok")
````

<details>
<summary><b>Uma escrita, duas funções</b></summary>

```python
exames[paciente] = valor
return exames
```

A mesma linha **cria** a chave que não existe e **troca** o valor da que existe. Não há um
comando para inserir e outro para atualizar — é o que o slide chamava de "a mesma escrita
faz as duas coisas".

E repare no que aconteceu: a sua função alterou o dicionário de quem chamou. Dicionário
não é copiado ao entrar numa função; lá dentro é o **mesmo** objeto.

</details>

<details>
<summary><b>Se veio um erro de tipo, foi a pegadinha do slide</b></summary>

`for chave in exames` entrega **chaves**, não valores. Somar chaves de texto dá
`TypeError: unsupported operand type(s) for +=: 'int' and 'str'`.

As duas saídas:

```python

    total += exames[chave]        total += valor
```

As duas estão certas. A segunda diz em voz alta o que você quer, e é a que se lê melhor
daqui a um mês.

</details>


Sem o `.get`, você precisaria de um `if item in saida: ... else: ...` — quatro linhas para

daquela chave.

````python
def soma_por_chave(pares):
    """Recebe `[(chave, valor), ...]` e devolve chave -> soma dos valores."""
````

````python
assert soma_por_chave([("a", 10), ("a", 20)])         == {"a": 30}
assert soma_por_chave([("a", 1), ("b", 4), ("a", 3)]) == {"a": 4, "b": 4}
assert soma_por_chave([])                             == {}

consultas = [("P001", 17), ("P002", 4), ("P001", 21)]
assert soma_por_chave(consultas) == {"P001": 38, "P002": 4}

print("ok")
````

<details>
<summary><b>Repare no cabeçalho do laço</b></summary>

```python

```

`pares` é uma lista de tuplas, e você desempacota cada uma direto no `for`. É a mesma
escrita de `for chave, valor in d.items()` — porque `.items()` também devolve tuplas.

E é a mesma linha do exercício 10, com `+ valor` no lugar de `+ 1`. Contar é somar de um em um.

> **Se fosse média**, você precisaria de **dois** dicionários em paralelo — `somas` e
> `quantos` — e dividiria um pelo outro no fim. Soma guarda um número por chave; média
> precisa de dois.

</details>

## Fecho · as suas funções conversando

Nenhuma destas funções sabe que as outras existem. A célula abaixo usa **seis delas
juntas** para responder uma pergunta que nenhuma responderia sozinha: *em que semana do ano
os pacientes do estudo estão pior?*

````python
pares = []
with open("dataset.csv") as arquivo:
    arquivo.readline()
    for linha in arquivo:
        quem, semana, humor, sono = linha.strip().split(",")
        pares.append((int(semana), float(humor)))

total = soma_por_chave(pares)                  # exercício 11

medias = {}
for semana in total:
    medias[semana] = round(total[semana] / 200, 2)   # 200 pacientes em toda semana

print("semanas medidas :", len(medias))
print("pontas da série :", pontas(list(medias.values())))        # exercício 6
print("seis primeiras  :", comeco(list(medias.values()), 6))     # exercício 1
print()

ruins = acima_de(medias, 14.0)                 # exercício 12
print("semanas com média acima de 14.0:", len(ruins), "de", len(medias))
print()

piores = []
for p in range(1, 21):
    semana, _ = pior_semana("dataset.csv", f"P{p:03d}")          # exercício 14
    piores.append(semana)

print("a pior semana de cada um dos 20 primeiros pacientes, contada:")
print(contagem(piores))                        # exercício 10
````

<details>
<summary><b>O que a última linha mostra</b></summary>

Um dicionário quase todo de **uns**: vinte pacientes, quase vinte semanas diferentes. **Não
existe** uma semana ruim comum — o dado é sorteado, então a "pior semana" de cada paciente
cai em qualquer lugar.

Isso é o resultado certo, e é um resultado que você só enxerga porque contou. Num estudo de
verdade, é exatamente essa contagem que separa um padrão real de uma coincidência — e é a
primeira pergunta que você vai fazer a qualquer modelo no NeuroAI.

</details>
