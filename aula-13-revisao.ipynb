### 4 · O laço que olha para o vizinho

Leia a célula abaixo e responda **antes de rodar**:

1. Por que o `range` começa em **1** e não em 0?
2. Por que não basta perguntar `sinal[i] > limiar`? O que a segunda metade do `and` impede?
3. Antes de rodar: quantos cruzamentos você conta à mão nesta lista?

```python
sinal = [2, 45, 50, 8, 60, 70, 3]
limiar = 40

n = 0
for i in range(1, len(sinal)):
    if sinal[i] > limiar and sinal[i - 1] <= limiar:
        n = n + 1

print(n)
```

### 5 · Duas funções quase iguais

Leia a célula abaixo e responda **antes de rodar**:

1. Qual é a única diferença entre as duas funções?
2. O que a segunda imprime? Por quê?
3. Para que serve o `if` das duas primeiras linhas da primeira função?

```python
def media(valores):
    if len(valores) == 0:
        return 0.0
    return sum(valores) / len(valores)


def media_sem_return(valores):
    sum(valores) / len(valores)


print(media([2, 4, 6]))
print(media_sem_return([2, 4, 6]))
print(media([]))
```

### 10 · Criar vazio e preencher

Leia a célula abaixo e responda **antes de rodar**:

1. O que `np.zeros(6)` cria, e por que criar isso antes do laço?
2. Por que o laço começa em 1?
3. O que aconteceria se começasse em 0? Daria erro?
4. A última linha não tem `if` nem laço. O que ela faz?

```python
v = np.zeros(6)
v[0] = 20

for i in range(1, 6):
    v[i] = v[i - 1] - 6

print(v)

v[v < 0] = 0
print(v)
```
